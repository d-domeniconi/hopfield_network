```python
import modules.hopfield_network as hp
import matplotlib.pyplot as plt
```

```python
img_tolkien, vetor_tolkien = hp.preprocessar_para_hopfield("dataset/tolkien.png", 
                                                           size = 100, 
                                                           threshold = 0.90,
                                                           swap_colors = True)
```

```python
plt.imshow(img_tolkien, cmap='gray', vmin=-1, vmax=1)
plt.axis('off')
plt.show()
```

```python
tolkien_wrong = hp.add_noise(vetor_tolkien, noise_level=0.25)
```

```python
hp.vizualize([tolkien_wrong],[],[], len(tolkien_wrong), save = False, recall = False)
```

```python
w = hp.train(len(vetor_tolkien), [vetor_tolkien])
```

```python
tolkien_corrected = hp.retrieve(len(vetor_tolkien), w, tolkien_wrong)
```

```python
hp.vizualize([tolkien_corrected],[],[], len(tolkien_wrong), save = False, recall = False)
```
